## 4. Separate Target and Input Variables

The target is `price`. All other columns are inputs.

**Hint:** Check the Data Dictionary to identify which columns are numeric, binary, and categorical.

```python
# Separate the target variable ('price') from the inputs for both train and test
# Example:
#   train_y = train['price']
#   train_inputs = train.drop('price', axis=1)

train_y = train['price']
train_y.head()

test_y = test['price']
test_y.tail()
```

```python
# seperate inputs from train and test

train_inputs = train.drop('price', axis=1)
train_inputs.head()

test_inputs = test.drop('price', axis =1)
test_inputs.head()
```

---
## 5. Identify Column Types

Look at the Data Dictionary and your data. Classify each input column as:
- **Numeric** (continuous values)
- **Binary** (0/1 values)
- **Categorical** (text categories)

Create three Python lists with the column names.

```python
# Define lists of column names by type


numeric_columns = ["bedrooms", "bathrooms", "sqft_living", "sqft_lot", "floors", "sqft_above", "sqft_basement", 
                        "lat", "long", "sqft_living15", "sqft_lot15", 
                                "condition", "grade", "yr_built", "yr_renovated"]   
print(numeric_columns)

binary_columns =["waterfront"]    

categorical_columns = ["view", "zipcode"] 
```

---
## 6. Build the Preprocessing Pipeline

Use `Pipeline` and `ColumnTransformer` from scikit-learn to handle:
- **Numeric columns:** impute missing values (median), then standardize
- **Categorical columns:** impute missing values ('unknown'), then one-hot encode
- **Binary columns:** impute missing values (most frequent)

Refer to Tutorial 2 for the pattern.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
# Define the numeric transformer (imputer + scaler)

numeric_transformer = Pipeline(steps=[
                ('imputer', SimpleImputer(strategy='median')),
                ('scaler', StandardScaler())])
```

```python
# Define the categorical transformer (imputer + one-hot encoder)

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='unknown')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))])
```

```python
# Define the binary transformer (imputer only)

binary_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent'))]) 
```

```python
# Combine into a ColumnTransformer called 'preprocessor

preprocessor = ColumnTransformer([
        ('num', numeric_transformer, numeric_columns),
        ('cat', categorical_transformer, categorical_columns),
        ('binary', binary_transformer, binary_columns)],
        remainder='drop')


```

---
## 7. Transform the Data

Remember: `fit_transform()` on **train**, `transform()` on **test**.

```python
# fit_transform the training inputs

train_x = preprocessor.fit_transform(train_inputs)

train_x
```

```python
# transform the test inputs
test_x = preprocessor.transform(test_inputs)

test_x
```

```python
#  Print the shape of train_x and test_x to verify
train_x.shape
```

```python

test_x.shape
```

## 8. Compute the Baseline

The baseline for a regression task is to predict the **mean** of the target for every observation. Compute the baseline RMSE so you have something to compare your model against.

```python
mean_value = train_y.mean()


from sklearn.metrics import mean_squared_error

#Compute the baseline RMSE

baseline_predictions = np.full(len(test_y), train_y.mean())
baseline_rmse = np.sqrt(mean_squared_error(test_y, baseline_predictions))

print(baseline_rmse)
```

---
## 9. Train a Linear Regression

As you have seen in previous classes, `LinearRegression` fits a linear relationship between the inputs and the target. Train it on the preprocessed training data, predict on the test data, and compute the RMSE.

Compare your model's RMSE to the baseline RMSE from the previous step.

```python
from sklearn.linear_model import LinearRegression

# Create a LinearRegression, fit it on train_x and train_y
lin_reg_model = LinearRegression()
lin_reg_model.fit(train_x, train_y)


```

```python
# Predict on test_x and compute RMSE (test)
# Generate predictions for the test set
test_preds = lin_reg_model.predict(test_x)

# Calculate the RMSE
lin_test_rmse = np.sqrt(mean_squared_error(test_y, test_preds))

print(lin_test_rmse)
```

```python
# Also compute RMSE on the training set (to check for overfitting)
train_preds = lin_reg_model.predict(train_x)

lin_train_rmse = np.sqrt(mean_squared_error(train_y, train_preds))

print(lin_train_rmse)
```

## 10. Train a Decision Tree

*We will cover Decision Trees in detail in the coming weeks.* For now, just treat it as a different model and compare the results. The code pattern is identical to Linear Regression — only the model class changes.

Train a `DecisionTreeRegressor` and compare its RMSE to the Linear Regression and the baseline. Pay attention to the difference between train and test RMSE — what does it tell you?

```python
from sklearn.tree import DecisionTreeRegressor

# Create a DecisionTreeRegressor (use random_state=42)
# Fit on train_x and train_y

tree_reg = DecisionTreeRegressor(random_state=42)
tree_reg.fit(train_x, train_y)
```

```python
# Predict on test_x and compute RMSE (test)

test_preds_tree = tree_reg.predict(test_x)
tree_test_rmse = np.sqrt(mean_squared_error(test_y, test_preds_tree))

print(tree_test_rmse)
```

```python
# compute RMSE on the training set (to check for overfitting)

train_preds_tree = tree_reg.predict(train_x)
tree_train_rmse = np.sqrt(mean_squared_error(train_y, train_preds_tree))

print(tree_train_rmse)
```

## 11. Train a Random Forest

#Train a `RandomForestRegressor` and compare its RMSE to the other models and the baseline.

```python
from sklearn.ensemble import RandomForestRegressor

# Create a RandomForestRegressor (use n_estimators=100, random_state=42)
# Fit on train_x, train_y

forest_reg = RandomForestRegressor(n_estimators=100, random_state=42)
forest_reg.fit(train_x, train_y)
```
